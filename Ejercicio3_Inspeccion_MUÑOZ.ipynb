## Ejercicio 3: Agente de inspección de calidad

### Ficha PEAS

**P (Medida de desempeño):** Detectar correctamente la cantidad de defectos y tomar una decisión adecuada sobre el producto.

**E (Entorno):** Matriz de inspección de 5x5 que representa las partes revisadas del producto.

**A (Actuadores):** Aprobar, enviar a revisión manual o rechazar el producto.

**S (Sensores):** Valores de la matriz donde 0 representa una parte normal y 1 una parte defectuosa.
