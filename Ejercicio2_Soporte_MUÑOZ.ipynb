## Ejercicio 2: Agente de soporte técnico

### Ficha PEAS



**P (Medida de desempeño):** Atender primero los tickets que necesitan mayor prioridad y evitar tiempos de espera muy largos.

**E (Entorno):** Sistema de soporte técnico donde llegan los tickets de los clientes.

**A (Actuadores):** Asignar prioridad alta, media o baja al ticket.

**S (Sensores):** Tiempo de espera en minutos, nivel de urgencia y si el cliente es premium.

### Justificación de las reglas

La prioridad del ticket se decide tomando en cuenta la urgencia, el tiempo de espera y si el cliente es premium.
Los casos con urgencia alta tendrán prioridad alta, al igual que los tickets que lleven demasiado tiempo esperando.
Si la urgencia es media o el cliente es premium se asignará prioridad media.
Los demás casos tendrán prioridad baja.

### Código

```python
# funcion para asignar la prioridad de un ticket
def agente_soporte(tiempo_espera_min, urgencia, es_cliente_premium):

    # si la urgencia es alta se le da prioridad alta
    if urgencia == "alta":
        accion = "Prioridad alta"
        motivo = "El ticket tiene urgencia alta."

    # si lleva mas de 30 minutos esperando tambien pasa a prioridad alta
    elif tiempo_espera_min > 30:
        accion = "Prioridad alta"
        motivo = "El ticket lleva mucho tiempo esperando."

    # si la urgencia es media se le da prioridad media
    elif urgencia == "media":
        accion = "Prioridad media"
        motivo = "El ticket tiene urgencia media."

    # si es cliente premium se le da prioridad media
    elif es_cliente_premium:
        accion = "Prioridad media"
        motivo = "El cliente es premium."

    # los demas casos quedan con prioridad baja
    else:
        accion = "Prioridad baja"
        motivo = "El ticket no necesita una prioridad mayor."

    # devolvemos la prioridad y el motivo
    return accion, motivo
```

### Simulación y pruebas

```python
# casos para probar el agente de soporte
casos = [
    (5, "baja", False),     # poca espera y urgencia baja
    (10, "media", False),   # urgencia media
    (5, "alta", False),     # urgencia alta
    (20, "baja", True),     # urgencia baja pero es premium
    (40, "baja", False),    # urgencia baja pero lleva mucho esperando
    (45, "baja", True),     # mucha espera y tambien es premium
    (35, "alta", True)      # cumple varias condiciones de prioridad alta
]

# recorremos todos los casos
for tiempo, urgencia, premium in casos:

    # enviamos los datos al agente
    accion, motivo = agente_soporte(tiempo, urgencia, premium)

    # mostramos los datos y el resultado
    print("tiempo de espera:", tiempo)
    print("urgencia:", urgencia)
    print("cliente premium:", premium)
    print("accion:", accion)
    print("motivo:", motivo)
    print("------------------------")
```
